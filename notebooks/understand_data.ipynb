## Understanding the Dataset

The plotting code is in `include/plots_data.py`. Every figure is also saved to `figures/` as a PDF (for the report) and a PNG.

```python
from include.data import load, load_hidden
from include.figures import use_paper_style, savefig
from include import plots_data as plots

use_paper_style()
```

### 1. Size of each dataset

```python
for name in [20, 30, 40, "40-adapt"]:
    X, y = load(name)
    print(f"{name}: {X.shape[0]} rows, {X.shape[1]} features, {y.mean():.1%} have y = 1, "
          f"{len(load_hidden(name))} hidden rows")
```

```python
fig = plots.plot_sizes()
savefig(fig, "data_sizes")
```

### 5. Features vs label
Here corr is $\mathrm{mean}(s \cdot t)$ with $t = 2y - 1$, so $(1 + |\mathrm{corr}|)/2$ is the accuracy you get by predicting $y$ from that product alone. The grey band is the noise level $2/\sqrt{N}$.

```python
fig = plots.plot_single_corr()
savefig(fig, "data_single_corr")
```

Products of two features.

```python
fig = plots.plot_pair_corr()
savefig(fig, "data_pair_corr")
```

Largest $|\mathrm{corr}|$ for single features, pairs and triples. Dotted lines are the same thing with the labels shuffled, i.e. what you'd see by chance.
